```python
cache += dx**2  # 그래디언트를 제곱해서 누적
# cache는 이 파라미터가 얼마나 크게 자주 움직여왔는지의 히스토리를 기록
x += -learning_rate * dx / (np.sqrt(cache) + 1e-7)
# cache가 크다 그래디어트가 컸다 -> 분모가 커짐 -> 학습률이 작아짐
# cache가 작다 그래디언크가 작았다 -> 분모가 작아짐 -> 학습률이 커짐
# 1e-7은 분모가 0되는걸 막는 안전장치
```
