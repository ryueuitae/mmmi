```python
import numpy as np
D = np.random.randn(1000,500) #입력 데이터 D 생성 1000개의 샘플 500차원 평균 0 표준편차 1
hidden_layer_sizes = [500]*10 #10개의 은닉층 각 층의 뉴런 수 500
nonlinearities = ['tanh']*len(hidden_layer_sizes) # 각 층에 적용할 활성화 함수 지정 tanh 가 10번 반복
```

```python
act = {'relu': lambda x:np.maximum(0,x),'tanh':lambda x:np.tanh(x)}
# relu 라는 이름표를 붙이면 ReLU 함수가 tanh 이름표를 붙이면
Hs = { }
for i in range(len(hidden_layer_sizes)): # i를 10번 반복
  X = D if i == 0 else Hs[i-1] # 지금이 몇번째 층이냐에 따라서 입력 다르게
  fan_in = X.shape[1] # 이 층에 들어오는 데이터 열의 개수 = 입력 뉴런 수
  fan_out = hidden_layer_sizes[i] # 이 층이 내보낼 뉴런 수
  w = np.random.randn(fan_in, fan_out) * 1 # 이 층의 가중치 행렬을 만듦

  H = np.dot(X,w) # 행렬곱 -> 선형 변환 근데 왜 이게 선형 변환이야?
  #-> 선형 변환의 정의와 행렬 곱이 일치함
  H = act[nonlinearities[i]](H)
  Hs[i] = H
```

```python
print ('input layer had mean %f and std %f' % (np.mean(D), np.std(D))) # D의 평균과 표준편차 출력
layer_means = [np.mean(H) for i,H in Hs.items()] # 10개 층 각각의 평균값 뽑아서 새리스트
layer_stds = [np.std(H) for i,H in Hs.items()] # 10개 층 각각의 표준편차 뽑아서 새리스트
for i,H in Hs.items(): # 각 층 순회 하면서 출력
  print ('hidden layer %d had mean %f and std %f' % (i+1,layer_means[i],layer_stds[i]))
```
